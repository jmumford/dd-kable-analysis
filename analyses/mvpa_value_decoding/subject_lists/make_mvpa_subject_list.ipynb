## (re)establishing code that was used to make the subject list
This list is just a copy of the subjects to use in the MVPA anlaysis.  I couldn't use the csv for some reason and just make a txt file for my slurm jobs.

```python
cfg = load_config()

subject_lists_dir = Path(cfg.subject_lists)

runs_csv = subject_lists_dir / 'initial_qa_pass_and_mask_pass_subjects_runs.csv'

test_list_path = subject_lists_dir / 'mvpa_subject_list_TEST.txt'

existing_list_path = subject_lists_dir / 'mvpa_subject_list.txt'


runs_csv, test_list_path, existing_list_path
```

```python
subject_ids = []
seen = set()

with runs_csv.open(newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        sub_id = row['sub_id'].strip()
        if sub_id and sub_id not in seen:
            seen.add(sub_id)
            subject_ids.append(sub_id)

test_list_path.write_text('\n'.join(subject_ids) + '\n')
print(f'Wrote {len(subject_ids)} subject IDs to {test_list_path}')
```

```python
test_lines = test_list_path.read_text().splitlines()
existing_lines = existing_list_path.read_text().splitlines()

print(f'TEST subjects: {len(test_lines)}')
print(f'Existing subjects: {len(existing_lines)}')
print(f'Exact match: {test_lines == existing_lines}')

if test_lines != existing_lines:
    only_in_test = sorted(set(test_lines) - set(existing_lines))
    only_in_existing = sorted(set(existing_lines) - set(test_lines))
    print('Only in TEST:', only_in_test[:10])
    print('Only in existing:', only_in_existing[:10])
```
